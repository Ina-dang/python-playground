```python
import numpy as np


def filter_readings(values):
    """-20 이상 80이하는 유효값
    유효값중 30이상은 경고대상

    기대 반환값
    (np.array([0, 29, 30, 80, -20]),
    np.array([30, 80]))
    """
    new_list = (values >= -20) & (values <= 80)
    valid_values = values[new_list]
    warn_list = valid_values >= 30
    alert_values = valid_values[warn_list]

    return valid_values, alert_values


values = np.array([0, 29, 30, 80, 81, -20, -21])
values = np.array([-20, 30, 80, 81])
values = np.array([35, 0, 35, -5])
filter_readings(values)
```

```python
import numpy as np


def filter_readings(values):
    """-20 이상 80이하는 유효값
    유효값중 30이상은 경고대상

    기대 반환값
    (np.array([0, 29, 30, 80, -20]),
    np.array([30, 80]))
    """
    new_list = (values >= -20) & (values <= 80)
    valid_values = values[new_list]
    warn_list = valid_values >= 30
    alert_values = valid_values[warn_list]

    return valid_values, alert_values


values = np.array([0, 29, 30, 80, 81, -20, -21])
values = np.array([-20, 30, 80, 81])
values = np.array([35, 0, 35, -5])
values = np.array([-21, 81])
values = np.array([])
values = np.array([29.5, 30.0])
filter_readings(values)
```

## 제출 기록 보완

기존 함수와 설명은 그대로 두고, B 문제의 dtype 및 교재 2쪽·7쪽 입력별 실제 실행 결과를 기록했습니다.


```python
# B 경로 확인 문제 1: 기존 풀이의 마스크와 dtype 확인
a = np.array([3, 8, 5, 11])
m = (a > 4) & (a < 10)
picked = a[m]
print("m:", m)
print("type(m):", type(m))
print("m.dtype:", m.dtype)
print("picked:", picked)
assert m.dtype == np.dtype(bool)
```

```python
# 교재 2쪽 예시와 7쪽의 다섯 입력을 각각 실행합니다.
# 아래 반복문은 테스트 실행용이며, filter_readings 함수는 변경하지 않았습니다.
test_cases = [
    ("2쪽 예시", np.array([0, 29, 30, 80, 81, -20, -21]), [0, 29, 30, 80, -20], [30, 80]),
    ("경계값", np.array([-20, 30, 80, 81]), [-20, 30, 80], [30, 80]),
    ("순서와 중복", np.array([35, 0, 35, -5]), [35, 0, 35, -5], [35, 35]),
    ("모두 범위 밖", np.array([-21, 81]), [], []),
    ("빈 입력", np.array([], dtype=int), [], []),
    ("소수와 경고 경계", np.array([29.5, 30.0]), [29.5, 30.0], [30.0]),
]
for case_name, case_values, expected_valid, expected_alert in test_cases:
    input_before = case_values.copy()
    result = filter_readings(case_values)
    assert isinstance(result, tuple) and len(result) == 2
    assert isinstance(result[0], np.ndarray) and isinstance(result[1], np.ndarray)
    assert result[0].ndim == result[1].ndim == 1
    assert result[0].dtype == result[1].dtype == case_values.dtype
    valid_matches = np.array_equal(result[0], np.array(expected_valid, dtype=case_values.dtype))
    alert_matches = np.array_equal(result[1], np.array(expected_alert, dtype=case_values.dtype))
    input_unchanged = np.array_equal(case_values, input_before)
    assert valid_matches and alert_matches and input_unchanged
    print(f"[{case_name}]")
    print("입력:", repr(case_values))
    print("반환:", repr(result))
    print("바깥 타입:", type(result))
    print("안쪽 타입:", type(result[0]), type(result[1]))
    print("dtype:", result[0].dtype, result[1].dtype)
    print("shape:", result[0].shape, result[1].shape)
    print("값 일치:", valid_matches, alert_matches, "/ 입력 보존:", input_unchanged)
    print()
print("교재 6개 입력 검사 통과")
```
