```python
import numpy as np
```

```python
probe = np.array([[3, 30], [5, 50], [7, 70]])
```

- 두 센서 각 각의 결과니까 결과는 2개가 나와야하고, 첫 센서 평균은 각 배열의 0번째 값으로 계산한다


```python
probe_sensor_a_avg = sum(probe[:, 0]) / len(probe[:, 0])
probe_sensor_b_avg = sum(probe[:, 1]) / len(probe[:, 1])
probe_sensor_a_avg, probe_sensor_b_avg
```

```python
samples = np.array([[2, 10], [4, 20], [6, 30]])
print(samples.shape)
```

```python
sensor_means = np.mean(samples, axis=0)
print(sensor_means)
print(sensor_means.shape)
```

```python
overall_mean = np.mean(samples)
print(overall_mean)
```

```python
readings = np.array(
    [[4, 20], [8, 40], [12, 60], [16, 80]],
)
before = readings.copy()


def sensor_average(readings):
    if readings.shape[0] > 0:
        return np.mean(readings, axis=0)


after = sensor_average(before)
print(after)
print(after.shape)
print(np.array_equal(before, readings))
```

- 측정 횟수인 행 축을 모아 센서별 평균을 남겼다


```python
data = np.array([[5, 15], [10, 30], [15, 45]])
wrong = np.mean(data, axis=1)  # 센서 별 계산을 원하면 axis 0을 해야함
print(wrong.shape)
```

```python
data = np.array([[5, 15], [10, 30], [15, 45]])
correct = np.mean(data, axis=0)
print(correct.shape)
```
