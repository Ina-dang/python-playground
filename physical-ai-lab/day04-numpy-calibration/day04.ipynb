```python
import numpy as np

records = np.array([[101, 28], [102, 31]])
records_copy = records.copy()
fix_data = records_copy[:, 1] + 2
records_copy[:, 1] = fix_data

records_copy
```

```python
def calibrate_temperatures(records, offset=2):
    records_copy = records.copy()
    fix_data = records_copy[:, 1] + offset
    records_copy[:, 1] = fix_data
    return records_copy


calibrate_temperatures(records, offset=2)
```

```python
before = records.copy()
result = calibrate_temperatures(records, 2)
print(np.array_equal(records, before))  # 기대: True
print(result.dtype == records.dtype)  # 기대: True
print(result is records)
```
