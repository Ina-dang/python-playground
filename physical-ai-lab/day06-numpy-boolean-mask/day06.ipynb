```python
# A
records = np.array([[14, 29], [82, 30], [37, 34], [6, -2], [51, 30]], dtype=np.int64)
limit = 30


def select_hot_sensors(records, limit):
    # 1. 온도가 limit 이상인 행만 반환해요.
    # 2. 결과는 NumPy 배열이며 [센서 번호, 온도] 두 열을 유지해요.
    # 3. 남은 행은 입력에서 등장한 순서를 유지해요. 정렬하지 않아요.
    # 4. 입력 records의 값은 바꾸지 않아요.
    # 5. 선택된 행이 없어도 결과 shape는 (0, 2)예요.
    # 6. 이번에는 for문 대신 불리언 마스크를 사용해요.
    temp = records[:, 1] >= limit
    return records[temp]


before = records.copy()
result = select_hot_sensors(records, limit)
print(result)
print(result.shape)
print(np.array_equal(records, before))
```

```python
# B
cases = {
    "B1": np.array([[41, 9], [42, 10], [43, 11]], dtype=np.int64),
    "B2": np.array([[55, -3], [56, 4]], dtype=np.int64),
    "B3": np.array([[91, 15], [12, 10]], dtype=np.int64),
}
limit = 10  # 세 경우에 공통으로 적용

for key, value in cases.items():
    result = select_hot_sensors(value, limit)
    print(f"result {key}\n {result}\n{result.shape}\n{'-'*10}")
```

```python
# C
sample = np.array([[44, 16], [19, 21], [73, 25]])
values = sample[:, 1]
print(values)
flag = values >= 21
wrong = values[flag]
print("before:", wrong, wrong.shape)
```

C-설명: 기존 코드는 온도 데이터만 선택하므로, 온도 21과 25에 대응하는 센서 번호 19와 73이 결과에서 누락된다.
C-이유: 센서 번호와 온도의 대응 관계를 유지하려면, 불리언 마스크 flag를 원본 2차원 배열 sample에 적용하여 조건을 만족하는 행 전체를 선택해야 한다.


```python
# 온도가 21 이상인 센서의 번호와 온도를 두 열짜리 배열로 반환

sample = np.array([[44, 16], [19, 21], [73, 25]])
values = sample[:, 1]
flag = values >= 21
fixed = sample[flag]
print("after:", fixed, fixed.shape)
```
