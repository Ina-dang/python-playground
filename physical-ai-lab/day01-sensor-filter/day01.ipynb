# 값 구분하기
- 규칙: -20~80은 유효값(양 끝 포함). 유효값 중 30 이상은 경고 대상.

```python
sample_list = [22, None, 0, 31, -99, 30, 85]
possible_list = []
warn_list = []
for item in sample_list:
    if item == None: 
        continue
    if -20 <= item <= 80:
        possible_list.append(item)
    else:
        warn_list.append(item)

print(f"유효값 리스트: {possible_list} \n경고값 리스트: {warn_list}")
```

- Q: None과 0을 같은 방식으로 처리해도 될까요?
- A: 안된다. 0은 측정할 수 있고 숫자로 처리되지만 none은 값이 없다 라는 의미를 표현하기 때문에 다르게 사용해야한다.

# 규칙을 바꿔도 쓸수 있는 함수 설계

```python
def filter_readings(
    readings, min_value = -20, max_value=80, alert_threshold = 30
):
  data_list = []
  alert_list = []
  if readings is None: readings
  elif len(readings) == 0: readings
  for reading in readings:
    if reading is None: continue
    if (min_value <= reading <= max_value):
      if alert_threshold <= reading :
        alert_list.append(reading)
      data_list.append(reading)
  return data_list, alert_list
```

| 확인할 것 | 입력 | 내 예상 | 실제 결과 |
| --- | --- | --- | --- |
| 경계 안과 밖 | `[-21, -20, 80, 81]` |[-20,80] | [-20,80]|
| 빠짐·0·경고 경계 | `[None, 0, 29, 30]` |[0,29,30] |[0,29,30] |
| 순서와 중복 | `[40, -3, 40, 12]` |[40, -3, 40, 12] | [40, -3, 40, 12]|
| 빈 입력 | `[]` | []| []|
| 모두 빠진 입력 | `[None, None]` |[] |[] |
| 모두 범위 밖인 입력 | `[-21, 81]` |[] | []|

```python
print(filter_readings([-21,-20,80,81]))
print(filter_readings([None, 0, 29, 30]))
print(filter_readings([40, -3, 40, 12]))
print(filter_readings([]))
print(filter_readings([None, None]))
print(filter_readings([-21, 81]))
```

- ① None과 0을 구분한 이유
  - 각 값의 역할이 다르기 때문. None은 없다라는 표시고 0은 숫자로 계산이 가능한 부분이다.

- ② 왜 유효성을 확인한 뒤에 경고 여부를 판단했는지
  - 숫자와 None은 비교할 수 없으므로 None을 먼저 처리해야한다

- ③ 경계값이나 빈 입력 테스트에서 확인한 점 한 가지
  - 빈 입력을 위해 값을 먼저 체크하기 전 길이도 확인을 해야한다
