```python
class SensorCalibrator:
    def __init__(self, offset, scale):
        self.offset = offset
        self.scale = scale

    def apply(self, raw):
        return (raw - self.offset) * self.scale
```

```python
registry = {"left": SensorCalibrator(1, 2), "right": SensorCalibrator(3, 1)}
raw_by_name = {"right": 11, "left": 8}
```

```python
name = "left"
sensor = registry[name]
raw = raw_by_name[name]
value = sensor.apply(raw)
print(value)
```

```python
def apply_named(registry, raw_by_name):
    new_value = registry.copy()
    for id, value in raw_by_name.items():
        print(id, value)
        new_value[id] = new_value[id].apply(value)
    return new_value


registry = {
    "left": SensorCalibrator(2, 2),
    "center": SensorCalibrator(1, 3),
    "right": SensorCalibrator(4, 1),
}
packet_a = {"right": 12, "left": 9, "center": 7}
packet_b = {"center": 7, "right": 12, "left": 9}

result_a = apply_named(registry, packet_a)
result_b = apply_named(registry, packet_b)
print(result_a)
print(result_b)  # 키로 찾아서 계산
print(result_a == result_b)
```
