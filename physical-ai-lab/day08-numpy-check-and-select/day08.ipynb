```python
import numpy as np
```

```python
probe = np.array([[61, 2], [72, 5], [83, 8]])
selected = probe[probe[:, 1] >= 5]
selected
```

- 원본보관: selected 실행 전 probe의 값을 .copy()로 복사한 후에 selected에서 사용하는 변수를 카피된 값으로 사용한다.
- 파이썬 and는 왼쪽 배열 전체를 boolean으로 평가하려고 하지만 배열에 여러 True, False 값이 있어서 ambigous 오류가 난다.


```python
raw = np.array([[91, 7], [92, 9]])
before = raw.copy()
chosen = raw[raw[:, 1] >= 8]
print(chosen)
print(np.array_equal(before, raw))
print(np.array_equal(raw, chosen))
```

```python
demo = np.array([[11, 2], [22, 3], [33, 6]])
temp = demo[:, 1]
left = temp >= 3
right = temp < 6
mask = left & right
print(mask)
print(demo[mask])
```

```python
def pick_range(records, low, high):
    records_range = records[:, 1]
    left = records_range >= low
    right = records_range < high
    mask = left & right
    return records[mask]
```

```python
sample = np.array(
    [
        [104, 4],
        [215, 5],
        [326, 9],
        [437, 10],
    ],
)
low, high = 5, 10
```

```python
copy_sample = sample.copy()
result = pick_range(sample, low, high)
copy_result = pick_range(copy_sample, low, high)
print(copy_result)
print(copy_result.shape)
print(np.array_equal(sample, copy_sample))
```

```python
extra = np.array(
    [
        [58, 12],
        [69, 15],
        [70, 18],
    ],
)
low_b, high_b = 12, 18
```

```python
copy_extra = extra.copy()
result = pick_range(extra, low_b, high_b)
copy_result = pick_range(copy_extra, low_b, high_b)

print(copy_extra)
print(copy_extra.shape)
print(np.array_equal(extra, copy_extra))
```
