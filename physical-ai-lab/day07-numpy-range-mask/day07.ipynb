```python
import numpy as np
```

```python
demo = np.array([[901, 4], [902, 5], [903, 7], [904, 9], [905, 11]])
temps = demo[:, 1]
print(temps)
```

```python
left = temps >= 5
right = temps < 9
mask = left & right

print(left)
print(right)
print(mask)
```

```python
picked = demo[mask]
print(picked)
print(picked.shape)
```

```python
records = np.array([[104, 18], [102, 20], [107, 26], [101, 30], [109, 33]])


def pick_range(records, low, high):
    records_range = records[:, 1]
    left = records_range >= low
    right = records_range < high
    mask = left & right
    return records[mask]


print(records.shape)
result = pick_range(records, 20, 30)
print(result)
print(np.array_equal(records, result))
print(result.shape)

# 우사기 보완
# A-설명: 선택된 행은 온도가 low 이상이고 high 미만이라는 두 조건을 모두 만족한다. 아래 경계는 포함하고 위 경계는 제외한다.
# 위 False는 입력과 선택 결과가 다르다는 뜻이며, 원본 보존 검사는 B에서 실행 전후 입력끼리 비교한다.
```

- B1-관찰 (우사기 보완): 아래 경계 10과 같은 값은 포함되어 번호 42가 남고, 위 경계 15와 같은 값은 제외되어 번호 24는 남지 않는다.
- B2-관 찰 : 선 택 된 행 수 는 0개 / shape는 (0,2)
- B3-관 찰 : 결 과 의 센 서 번 호 순 서 는 97, 38


```python
# B
cases = {
    "B1": np.array([[61, 9], [42, 10], [73, 14], [24, 15]]),
    "B2": np.array([[85, 2], [16, 18]]),
    "B3": np.array([[97, 14], [38, 10]]),
}
low, high = 10, 15

for key, value in cases.items():
    # 우사기 보완: 각 사례의 실행 전후 입력을 비교한다.
    before = value.copy()
    result = pick_range(value, low, high)
    print(f"-----{key}--------")
    print(value.shape)
    print(result)
    print(np.array_equal(before, value))
    print(result.shape)
    print("------------------")
```

- C-원인 (우사기 보완): Python and는 왼쪽 배열 전체를 하나의 True 또는 False로 판단하려 한다. 그런데 배열에 여러 참·거짓 값이 있어 하나로 판단할 수 없으므로 오류가 난다.
- C-수 정 : 각 원 소 끼 리 두 조 건 을 합 치 기 위 해 & 기호를 사 용 했 다 .


```python
# C
probe = np.array([[54, 3], [17, 6], [88, 8]])
temps = probe[:, 1]
correct = (temps >= 4) & (temps < 8)
# 우사기 보완: 같은 위치의 두 조건을 &로 합치고 선택된 전체 행도 출력한다.
print(probe[correct])
print(correct)
```
