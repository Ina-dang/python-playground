```python
class SensorCalibrator:
    def __init__(self, offset, scale):
        self.offset = offset
        self.scale = scale

    def apply(self, raw):
        return (raw - self.offset) * self.scale


def apply_named(registry, raw_by_name):
    new_value = registry.copy()
    for name, value in raw_by_name.items():
        new_value[name] = new_value[name].apply(value)
    return new_value
```

```python
registry = {"left": SensorCalibrator(2, 2), "right": SensorCalibrator(1, 3)}
frames = [
    {"t_ms": 0, "raw": {"right": 7, "left": 10}},
    {"t_ms": 100, "raw": {"left": 12, "right": 8}},
    {"t_ms": 200, "raw": {"right": 6, "left": 11}},
]
```

```python
def record_frames(registry, frames):
    result = []

    # row를 apply_name로 계산하고
    for frame in frames:
        data = {}
        for id, value in frame.items():
            if id == "t_ms":
                data["t_ms"] = value
            if id == "raw":
                data["values"] = apply_named(registry, value)

        result.append(data)

    # {"t_ms": 해당 시간, "values": 보정값 사전} 로 담기
    return result


print(record_frames(registry, frames))
```
