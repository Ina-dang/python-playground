```python
class SensorCalibrator:
    def __init__(self, offset, scale):
        self.offset = offset
        self.scale = scale

    def apply(self, raw):
        return (raw - self.offset) * self.scale
```

```python
sensors = [
    SensorCalibrator(1, 2),
    SensorCalibrator(3, 1),
]
raw_values = [8, 11]
results = []
```

```python
for sensor, raw in zip(sensors, raw_values):
    value = sensor.apply(raw)
    results.append(value)
print(results)
```

```python
def add_pairs(left, right):
    if len(left) != len(right):
        raise ValueError("length mismatch")
    result = []
    for a, b in zip(left, right):
        result.append(a + b)
    return result


print(add_pairs([1, 2], [10, 20]))
print(add_pairs([3, 4], [10, 20]))
```

```python
def apply_all(sensors, raw_values):
    result = []
    if len(sensors) != len(raw_values):
        raise ValueError("length mismatch")
    for a, b in zip(sensors, raw_values):
        value = a.apply(b)
        result.append(value)
    return result


print(apply_all(sensors, raw_values))
raw_values = [10, 8, 13]
print(apply_all(sensors, raw_values))
```
