```python
# 측정한 두 센서의 값.
raw = np.array([[12, 105], [22, 205], [32, 305]])

# 각 열에 대응
offset = np.array([2, 5])
scale = np.array([0.1, 0.01])
```

```python
corrected = raw - offset
print(corrected)
```

```python
converted = corrected * scale
print(converted)
print(converted.shape)
```

```python
readings = np.array([[15, 230], [25, 430], [35, 630]])
offset = np.array([5, 30])
scale = np.array([0.2, 0.01])
```

```python
before = readings.copy()
```

```python
def calibrate(readings, offset, scale):
    result = readings - offset
    result = result * scale
    return result
```

- offset의 첫값은 readings의 각 행의 첫 열에 적용된다.
- 먼저 빼고 나중에 곱하는 이유는 기준오차를 빼고 열별 비율을 곱해주는 연산순서를 지키기 위해서다


```python
print(calibrate(before, offset, scale))
print(readings.shape, before.shape)
```

```python
extra = np.array([[8, 80], [12, 120], [16, 160]])
wrong_offset = np.array([1, 2, 3])
extra - wrong_offset
```

- 주석의 계산이 왜 맞지 않는지 두 배열의 shape를 이용해 설명해요.
  - 행의 개수가 다르기 때문이다

- “첫 센서에서 1, 둘째 센서에서 2를 뺀다”는 목적에 맞게 offset만 고쳐 실행해요.


```python
extra = np.array([[8, 80], [12, 120], [16, 160]])
correct_offset = np.array([1, 2])
extra - correct_offset
```
