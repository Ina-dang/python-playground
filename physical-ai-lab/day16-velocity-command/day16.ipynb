```python
def action_to_speed(action, cruise_mps):
    return 0.0 if action == "STOP" else cruise_mps
```

```python
for action in ["STOP", "GO"]:
    print(action, action_to_speed(action, 0.2))

print("GO at zero:", action_to_speed("GO", 0.0))
```

```python
def choose_action(front_m, stop_m):
    if front_m <= stop_m:
        return "STOP"
    return "GO"
```

```python
records = [
    {"t_ms": 0, "front_m": 1.20},
    {"t_ms": 100, "front_m": 0.50},
    {"t_ms": 200, "front_m": 0.30},
]


def replay_speeds(records, stop_m, cruise_mps):
    result = []
    for row in records:
        action = choose_action(row["front_m"], stop_m)
        result.append(
            {
                "t_ms": row["t_ms"],
                "front_m": row["front_m"],
                "action": action,
                "speed_mps": action_to_speed(action, cruise_mps),
            }
        )
    return result
```

```python
from copy import deepcopy

before = deepcopy(records)

for cruise_mps in [0.2, 0.4]:
    print("cruise_mps = ", cruise_mps)
    for row in replay_speeds(records, 0.5, cruise_mps):
        print(row)

assert records == before
```

- t_ms 0은 GO, 이후는 speed_mps와 action이 0.0와 STOP으로 액션이 달라졌다
