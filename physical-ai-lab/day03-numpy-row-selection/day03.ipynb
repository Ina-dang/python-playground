```python
import numpy as np
```

```python
boxes = np.array([[201, 12], [202, 25], [203, 19]])
weights = boxes[:, 1]
print(weights)  # [12 25 19]
print(weights.shape)  # (3,)
keep = weights >= 20
print(keep)  # [False  True False]
```

```python
picked = boxes[keep]
print(picked)
print(picked.shape)
```

```python
def select_hot_rows(records, threshold=30):
    # 1. 검사할 온도 열을 꺼내세요.
    picked = records[:, 1]
    # 2. 행마다 통과 여부를 만드세요.
    result = picked >= threshold
    # 3. 해당 행 전체를 반환하세요.
    return records[result]


records = np.array([[101, 22], [102, 35], [103, 28], [104, 30], [105, -5]])
threshold = 30
# expected: [[102, 35], [104, 30]]
# shape: (2, 2), dtype: records.dtype

select_hot_rows(records, threshold)
```

```python
empty = np.empty((0, 2), dtype=int)
# np.array([]) has shape (0,), not (0, 2)
empty.shape
# ? (0,2인데..?)
```

```python
print(select_hot_rows(np.array([[1, 29], [2, 30], [3, 31]]), 30))
print(select_hot_rows(np.array([[1, 10], [2, 20]]), 30))
print(select_hot_rows(empty, 30))
print(select_hot_rows(np.array([[7, 50], [7, 60]]), 50))
print(select_hot_rows(np.array([[1, -5], [2, 0]]), -5))
```

**질문 1:** 조건에 맞는 기록이 한 줄뿐이어도, 왜 (1, 2) 모양으로 남겨 둘까요?

**답:** 한 줄만 남아도 센서 번호와 온도는 각각 한 칸씩 있어야 하니까요. 두 열짜리 표로 남겨 두면, 다음에도 똑같이 온도 열을 꺼내서 쓸 수 있어요.

**질문 2:** 기록이 다섯 줄이라면, 남길지 말지 표시하는 True/False는 몇 개가 필요할까요?

**답:** 다섯 개요. 기록 한 줄마다 남길지 말지 하나씩 정해야 하니까요.

**질문 3:** 조건에 맞는 기록이 하나도 없을 때도, 왜 (0, 2) 모양을 유지할까요?

**답:** 기록은 없어도 센서 번호와 온도를 담는 두 열은 남겨 두려고요. 그러면 온도 열을 꺼낼 때 오류 대신 빈 결과가 나와서, 기록이 없을 때만 따로 코드를 바꾸지 않아도 돼요.
