```python
class Sensor:
    def __init__(self, offset):
        self.offset = offset

    def correct(self, raw):
        return raw - self.offset
```

```python
left = Sensor(1)
right = Sensor(4)

print(left.offset, right.offset)
```

```python
left.offset = 6
print(left.offset, right.offset)
```

```python
print(left.correct(9), right.correct(9))
```

```python
class SensorCalibrator:
    def __init__(self, offset, scale):
        self.offset = offset
        self.scale = scale

    def apply(self, raw):
        return (raw - self.offset) * self.scale
```

```python
a = SensorCalibrator(2, 3)
b = SensorCalibrator(4, 2)
```

```python
print(a.apply(10), b.apply(10))
```

```python
a.offset = 1
print(a.apply(10), b.apply(10))
```

- a.calibrate(10)을 호출하면 self는 a 객체를 가리킨다. a의 offset은 1, scale은 3이므로 (10−1)×3 = 27이 된다.
- b는 별도로 만든 객체라 a의 설정을 바꿔도 영향을 받지 않고, 결과는 12로 유지된다.


```python
a = SensorCalibrator(2, 3)
b = a
b.offset = 7
print(a.offset)

# b를 a와 같은 인스턴스를 써서 값이 덮어씌워진다
```
