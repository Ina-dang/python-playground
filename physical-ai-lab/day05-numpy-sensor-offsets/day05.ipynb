```python
import numpy as np
```

```python
temps = np.array([11, 24], dtype=np.int64)
changes = np.array([2, -3], dtype=np.int64)
new_temps = temps + changes
print(new_temps)  # [13 21]
print(new_temps.shape)  # (2,)
```

```python
demo = np.array([[4, 11], [8, 24]], dtype=np.int64)
print(demo[:, 1])
print(demo[:, 1].shape)  # 1차원으로 꺼냄
print(demo[:, 1:2])
print(demo[:, 1:2].shape)  # 2차원으로 꺼냄
```

```python
guess = np.array([6, 9], dtype=np.int64)
actual = np.array([6, 9], dtype=np.int64)
print(guess == actual)  # [True, True]
print(np.array_equal(guess, actual))  # True
```

```python
records = np.array([[61, 27], [12, 18], [48, 35], [5, -3]], dtype=np.int64)
offsets = np.array([-2, 3, 0, 1], dtype=np.int64)


def calibrate_per_sensor(records, offsets):
    records_copy = records.copy()
    temp = records_copy[:, 1] + offsets
    records_copy[:, 1] = temp
    return records_copy


print(calibrate_per_sensor(records, offsets))
```

```python
records_b1 = np.array([[7, -6]], dtype=np.int64)
offsets_b1 = np.array([4], dtype=np.int64)

print(calibrate_per_sensor(records_b1, offsets_b1))
```

```python
records_b2 = np.array([[9, 0], [2, 22]], dtype=np.int64)
offsets_b2 = np.array([0, 0], dtype=np.int64)
print(calibrate_per_sensor(records_b2, offsets_b2))
```

```python
records_b3 = np.array([[12, 18], [61, 27], [48, 35], [5, -3]], dtype=np.int64)
offsets_b3 = np.array([-3, 0, 1, 2], dtype=np.int64)
print(calibrate_per_sensor(records_b3, offsets_b3))
```

```python
probe = np.array([[15, 10], [26, 20], [37, 30]], dtype=np.int64)
steps = np.array([1, -2, 3], dtype=np.int64)
wrong = probe[:, 1:2] + steps
print(probe[:, 1:2].shape, steps.shape, wrong.shape)
print(wrong)

# 배열 모양이 다르다
```

```python
correct = probe[:, 1] + steps
print(probe[:, 1].shape, steps.shape, wrong.shape)
print(correct)
```
